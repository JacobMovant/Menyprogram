Tanken här är alltså att skapa en lista, väljer att köra på det uppenbara exemplet som är en handelslista, där man både kan visa listan, lägga till och ta bort, samt avsluta programmet. ## Pseudo-Kod / Planering

Start -
Start och någon loop-shenanigans, printa upp alternativ från följande funktioner.

Funktion 1 -
Visa listan. Printa bara ut listan som den är.

Funktion 2 - 
Lägga till grejer i listan, med variabel-info för said saker.

Funktion 3 -
Ta bort saker från listan. 

Funktion 4 - 
Avsluta programmet, enkelt liten "good night world" med ett "break" kommando.

END

```python
shopping_list = []
while True: # Funktion för hela meny UI't så att säga.
    print (" MENU ".center(20, "-")) # .center med informationen på hur det centreras inom parantes brevid.
    print ("1. Show list.")
    print ("2. Add to list.")
    print ("3. Remove from list.")
    print ("4. Save and Exit.")
    print ("-".center(20, "-")) # Bottom-banner för att snygga till det lite.

    option = input("Input a number.(1-4) ") # Input som bestämmer vad för alternativ man får från ovanstående.

    if option == "1": # Visning av listan med alternativ 'else' om den är tom.
        if len(shopping_list) == 0:
            print("Your list is empty!")
        else:
            print("\nShopping List:")
            for item in shopping_list:
                print("-", item)
                
    elif option == "2": # Tillägning till listan.
        item = input("What would you like to add to your list?")
        shopping_list.append(item)
        print(f"'{item}' has been added to your list.")

    elif option == "3": # Bortagning från listan.
        item = input("What would you like to remove to your list?")

        if item in shopping_list:
            shopping_list.remove(item)
            print (f"'{item} has been removed.'")

    elif option == "4": # Exit.
        print ("Exiting program. Happy shopping!")
        break
```
